```python
 # Import
import pandas as pd
import numpy as np
df = pd.read_csv("data\All_Data_Sains.csv")

 # Checking values
df.shape
```

```python
 # Checking for NULL values (Run again to test)
df.isnull().sum()
```

```python
 # Deleting NULL values
 df = df.dropna()
```

**CHECKING** largest and smallest values

```python
df.nsmallest(10, "prices_(£)")
```

```python
df.nlargest(10, "prices_(£)")
```

```python
 #Quick anomaly check
 df["prices_unit_(£)"].describe()
```

**CLEANING DATA**

```python
# REMOVING prices 0 or below

df = df[(df["prices_(£)"] > 0)]
```

FILTERING PRICES above/ below 50%

```python
# Sorting values and getting previous product price
df = df.sort_values(["names", "date"])

df["previous_price"] = df.groupby("names")["prices_(£)"].shift(1)

df["previous_price"]
```

```python
# Price change calculator (Percentage)

df["price_change"] = df["prices_(£)"] - df["previous_price"]

df["percent_change"] = (
    (df["prices_(£)"] - df["previous_price"])
    / df["previous_price"]
) * 100

df.sort_values(by="percent_change", ascending=False)
```

```python
 # Checking for NA values 
 df.isnull().sum()
```

```python
 #Dropping NA values again
df = df.dropna()
```

```python
 # Finding suspicious values 
suspicious = df[df["percent_change"].abs() > 50]

suspicious.sort_values(by='percent_change', ascending=True)
```

```python
df.shape
```

```python
 # REMOVING suspicious values 
suspicious = df[df["percent_change"].abs() > 50]

df = df[df["percent_change"].abs() <= 50]
```

```python
df.shape
```

**REMOVING** price_unit values

```python
 # ADJUST PERCENTAGE CHANGE
THRESHOLD = 50 #Adjust accordingly

 # Sort data by product and date
df = df.sort_values(["names", "date"])

 # Get the previous unit price for each product
df["previous_unit_price"] = (
    df.groupby("names")["prices_unit_(£)"]
      .shift(1)
)

 # Calculate the percentage change
df["pct_change"] = (
    abs(
        (df["prices_unit_(£)"] - df["previous_unit_price"])
        / df["previous_unit_price"]
    ) * 100
)

 # Find suspicious price changes
suspicious2 = df[df["pct_change"] > THRESHOLD]

 # Show the largest changes first
suspicious2 = suspicious2.sort_values(
    by="pct_change",
    ascending=False
)

 #Display
print(f"Rows flagged: {len(suspicious2)}")
display(suspicious2)
```

```python
 #Removing values
clean_df = df[
    (df["pct_change"] <= THRESHOLD) |
    (df["pct_change"].isna())
]

#Check results
print(f"Original rows: {len(df)}")
print(f"Clean rows: {len(clean_df)}")
print(f"Rows removed: {len(df) - len(clean_df)}")
```

```python
 clean_df.isnull().sum() #Checking for NA in new clean_df
```

```python
clean_df = clean_df.dropna() #Dropping NA in clean_df
```

**REMOVING** zero's and high values

```python
clean_df.shape # Check values
```

```python
 #Removing values
clean_df = clean_df[(df["prices_unit_(£)"] > 0.1) & (df["prices_unit_(£)"] < 100)]

print(f"Original rows: {len(df)}")
print(f"Clean rows: {len(clean_df)}")
print(f"Rows removed: {len(df) - len(clean_df)}")
```

FINAL CHECK FOR ANOMALIES

```python
clean_df.nlargest(10, "pct_change")
```

REMOVING CATEGORY

```python
final_df = clean_df[
    ~clean_df["category"].isin(["baby_products", "free-from"])]

print(f"Original rows: {len(clean_df)}")
print(f"Final rows: {len(final_df)}")
print(f"Rows removed: {len(clean_df) - len(final_df)}")
```

CLEANUP

```python
 # Renaming column
final_df = final_df.rename(columns={
    'names': 'product',
    'prices_(£)': 'prices',
    'prices_unit_(£)': 'prices_unit'
})
```

```python
 # Dropping unecessary columns
final_df = final_df.drop(columns=['previous_price', 'price_change', 'percent_change', 'previous_unit_price', 'pct_change'])
```

```python
 # Organizing Column order
final_df = final_df[['category', 'prices', 'prices_unit', 'unit', 'product', 'date', 'own_brand', 'supermarket']]
```
